```python
from sklearn.neighbors import NearestNeighbors
import tkinter as tk
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer


# Load the Netflix dataset
netflix_df = pd.read_csv('netflix_titles.csv')

# Preprocess the dataset
tfidf = TfidfVectorizer(stop_words='english')
netflix_df['description'] = netflix_df['description'].fillna('')
tfidf_matrix = tfidf.fit_transform(netflix_df['description'])

# Train the KNN algorithm on the preprocessed dataset
knn = NearestNeighbors(n_neighbors=10, algorithm='brute', metric='cosine')
knn.fit(tfidf_matrix)

# Define the GUI window and input fields
window = tk.Tk()
window.title("Netflix Movie Recommender")
window.geometry("500x400")
window.configure(bg='#A9CCE3')

title_label = tk.Label(window, text="Enter a movie title:", bg='#FA8072')
title_label.config(font=("Times New Roman", 16))
title_label.pack(pady=20)

title_entry = tk.Entry(window, width=40, font=("Arial", 14))
title_entry.pack(pady=10)

# Define the prediction function and output label
def predict():
    title = title_entry.get()
    idx = netflix_df[netflix_df['title'] == title].index[0]
    distances, indices = knn.kneighbors(tfidf_matrix[idx])
    recommendations = netflix_df.iloc[indices[0]]['title'].tolist()
    output_label.config(text="Recommended Movies: \n" + '\n'.join(recommendations))

output_label = tk.Label(window, wraplength=400, bg='#F5F5F5')
output_label.config(font=("Arial", 14))
output_label.pack(pady=20)

# Add the input fields and predict button to the window
predict_button = tk.Button(window, text="Get Recommendations", command=predict, bg='#FF4500', fg='#FFFFFF')
predict_button.config(font=("Arial", 14))
predict_button.pack(pady=20)

# Add a logo image to the window
logo_image = tk.PhotoImage(file="LOGO (1).png")
logo_label = tk.Label(window, image=logo_image, bg='#F5F5F5')
logo_label.image = logo_image # keep a reference to prevent garbage collection
logo_label.pack(side='bottom', pady=10)

# Start the GUI main loop
window.mainloop()
```
